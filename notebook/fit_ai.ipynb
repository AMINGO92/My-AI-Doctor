```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(mydata["Steps"], mydata["Weight"])

plt.xlabel("Steps")
plt.ylabel("Weight")
plt.title("🔥 Steps vs Weight Analysis")

plt.grid()

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(mydata["Calories_Intake"], mydata["Weight"], alpha=0.6)

plt.xlabel("Calories Intake")
plt.ylabel("Weight")
plt.title("🔥 Calories Intake vs Weight Analysis")

plt.grid()

plt.show()
```

```python
print("===== AI HEALTH ANALYSIS =====")

# Steps
if mydata["Steps"].mean() < 7000:
    print("🚶 Steps कमी आहेत (Target: 8000+)")
else:
    print("✅ Steps चांगले आहेत")

# Calories
if mydata["Calories_Intake"].mean() > 2000:
    print("❌ Calories जास्त आहेत (Reduce needed)")
else:
    print("✅ Calories control मध्ये आहेत")

# Sleep
if mydata["Sleep_Hours"].mean() < 7:
    print("😴 Sleep कमी आहे (Target: 7-8 hrs)")
else:
    print("✅ Sleep proper आहे")

# Weight Trend
if mydata["Weight"].iloc[-1] < mydata["Weight"].iloc[0]:
    print("🔥 Fat loss चालू आहे")
else:
    print("⚠️ Improvement needed")
```

```python
corr = mydata.corr(numeric_only=True)

print("=== CORRELATION MATRIX ===")
print(corr)
```

```python
from sklearn.linear_model import LinearRegression

# Model training
X = mydata[["Steps", "Calories_Intake", "Sleep_Hours"]]
y = mydata["Weight"]

model = LinearRegression()
model.fit(X, y)

print("Model Ready 🔥")
```

```python
# Test input (example)
new_data = [[7000, 1800, 7]]

prediction = model.predict(new_data)

print("👉 Predicted Weight:", prediction[0])
```

```python
# New data (तुझा future plan 🔥)
new_data = [[7000, 1800, 7]]

# Prediction
predicted_weight = model.predict(new_data)

print("Predicted Weight:", predicted_weight)
```

```python
if prediction[0] < 75.5:
    print("🔥 Weight कमी होत आहे")
else:
    print("⚠️ Weight control करायला पाहिजे")
```

```python
print(model.predict(pd.DataFrame({
    "Steps": [3000],
    "Calories_Intake": [2300],
    "Sleep_Hours": [5]
})))

print(model.predict(pd.DataFrame({
    "Steps": [8000],
    "Calories_Intake": [1700],
    "Sleep_Hours": [7.5]
})))
```

```python
from sklearn.metrics import r2_score

y_pred = model.predict(X)

print("Accuracy:", r2_score(y, y_pred))
```

```python
import matplotlib.pyplot as plt

plt.scatter(y, y_pred)

plt.xlabel("Actual Weight")
plt.ylabel("Predicted Weight")
plt.title("Actual vs Predicted")

plt.show()
```
